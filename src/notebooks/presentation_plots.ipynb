```python
import numpy as np
import matplotlib.pyplot as plt

import sys
sys.path.append('F:\\OneDrive - Telecom Business Transformers BV\\School\\Uni\\MSc Year 2\\Master Research Project\\Project\\masterproject\\src')
from plotting_functions import set_styles
from astrometry_equations import error_ellipses
set_styles()
```

```python
# FIGURE 1: Only a Hipparcos star
fig, ax = plt.subplots(1,1,figsize=(6,6))

hip_mean = np.array([0,0])
hip_pm = np.array([1, 1.5])
hip_covar = np.array([9, 6, 2])
hip_ellipse = error_ellipses(hip_mean, hip_covar, [1], fill=False, ec='black')

ax.scatter(*hip_mean, c='black', s=250, marker='*')
ax.arrow(*hip_mean, *(hip_mean+hip_pm), head_width=0.25, head_length=0.2, lw=2, fc='black')
ax.add_patch(hip_ellipse[0])

ax.set_xlabel(r'$\Delta$ RA [mas]')
ax.set_ylabel(r'$\Delta$ Dec [mas]')
ax.set_xlim(-5,5)
ax.set_ylim(-5,5)

plt.savefig(fpath+'only_hipp.png', bbox_inches='tight')
```

```python
# Figure 5: Error Normalized Distance
# FIGURE 3: Backpropagate Gaia
fig, ax = plt.subplots(1,1,figsize=(6,6))

hip_mean = np.array([0,0])
hip_covar = np.array([9, 6, 2])
hip_ellipse = error_ellipses(hip_mean, hip_covar, [1], fill=False, ec='black', label='Hipparcos Uncertainty')

ax.scatter(*hip_mean, c='black', s=250, marker='*')
ax.add_patch(hip_ellipse[0])

gaia_mean = np.array([7, 8])
gaia_covar = np.array([4, 2, -2])
gaia_ellipse = error_ellipses(gaia_mean, gaia_covar, [1], fill=False, ec='C0', label='Gaia Uncertainty')

ax.scatter(*gaia_mean, c='C0', s=250, marker='*')
ax.add_patch(gaia_ellipse[0])

comb_ellipse = error_ellipses(hip_mean, hip_covar + gaia_covar, [1], fill=False, ec='orange', ls='--', label='Combined Uncertainty')
ax.add_patch(comb_ellipse[0])
ax.plot([hip_mean[0], gaia_mean[0]], [hip_mean[1], gaia_mean[1]], zorder=-1, c='red', label=f'Separation: 10.6 mas\nD: 3.43')

ax.set_xlabel(r'$\Delta$ RA [mas]')
ax.set_ylabel(r'$\Delta$ Dec [mas]')
#ax.set_xlim(-5,5)
#ax.set_ylim(-5,5)
plt.legend()
plt.savefig(fpath+'error_norm.png', bbox_inches='tight')
```
